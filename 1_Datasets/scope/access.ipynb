# Access Scope

The scope project consists of four zarr cubes:

- surface_carbonate_system_v2026_1.zarr
- global_open_ocean_doc.zarr
- global_coastal_doc.zarr
- global_carbon_biogeochemistry.zarr


## 1. surface_carbonate_system_v2026_1.zarr

This dataset describes ocean surface carbon chemistry  including CO₂ fugacity, dissolved inorganic carbon, alkalinity, pH, aragonite saturation, and associated uncertainties.

```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
url = "https://deep-esdl-public.s3.eu-central-1.amazonaws.com/SCOPE/mediterranean/surface_carbonate_system_v2026_1.zarr"
ds = xr.open_zarr(url, consolidated=True)
ds
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3))
ds.dic.isel(time=0).plot(ax=ax[0])
ds.fco2.isel(time=0).plot(ax=ax[1])
plt.tight_layout()
```

## 2. global_coastal_doc.zarr

Global Dissolved Organic Carbon (DOC) dataset, containing DOC concentrations (mg m-3) with per pixel uncertainty estimates gridded on geographic projection at 4 km spatial and monthly temporal resolution for 1998-2022. The DOC products were produced using the coastal DOC algorithm of Toming et al. (2026) and an updated version of the open ocean DOC algorithm of Laine et al. (2024). The dataset was generated as part of the European Space Agency (ESA) Satellite-based observations of Carbon in the Ocean: Pools, fluxes and Exchanges (SCOPE) project with support from the National Centre of Earth Observation (NCEO).

```python
url = "https://deep-esdl-public.s3.eu-central-1.amazonaws.com/SCOPE/mediterranean/global_coastal_doc.zarr"
ds = xr.open_zarr(url, consolidated=True)
ds
```

## 3. global_open_ocean_doc.zarr

Global ocean carbon and phytoplankton dataset containing satellite-derived products for Dissolved Organic Carbon (DOC), Phytoplankton Carbon (PC), Particulate Inorganic Carbon (PIC), Particulate Organic Carbon (POC), and Phytoplankton Primary Production (PP), where available. The products include corresponding uncertainty estimates where provided. The datasets are gridded on a geographic grid at nominal 4 km spatial and monthly temporal resolution for 1998-2022, subject to the temporal coverage and spatial resolution of the individual products. The products were developed using a range of published and project-specific algorithms. This datacube was generated as part of the European Space Agency (ESA) Satellite-based observations of Carbon in the Ocean: Pools, fluxes and Exchanges (SCOPE) project, with support from the National Centre of Earth Observation (NCEO).

```python
url = "https://deep-esdl-public.s3.eu-central-1.amazonaws.com/SCOPE/mediterranean/global_carbon_biogeochemistry.zarr"
ds = xr.open_zarr(url, consolidated=True)
ds
```

## 4. global_carbon_biogeochemistry.zarr

Global ocean carbon and phytoplankton dataset containing satellite-derived products for Dissolved Organic Carbon (DOC), Phytoplankton Carbon (PC), Particulate Inorganic Carbon (PIC), Particulate Organic Carbon (POC), and Phytoplankton Primary Production (PP), where available. The products include corresponding uncertainty estimates where provided. The datasets are gridded on a geographic grid at nominal 4 km spatial and monthly temporal resolution for 1998-2022, subject to the temporal coverage and spatial resolution of the individual products. The products were developed using a range of published and project-specific algorithms. This datacube was generated as part of the European Space Agency (ESA) Satellite-based observations of Carbon in the Ocean: Pools, fluxes and Exchanges (SCOPE) project, with support from the National Centre of Earth Observation (NCEO).

```python
url = "https://deep-esdl-public.s3.eu-central-1.amazonaws.com/SCOPE/mediterranean/global_carbon_biogeochemistry.zarr"
ds = xr.open_zarr(url, consolidated=True)
ds
```

```python
fig, ax = plt.subplots(3, 4, figsize=(20, 9))
_ax = ax.flatten()
for i, (name, da) in enumerate(ds.data_vars.items()):
    da.isel(time=0).plot(ax=_ax[i], robust=True)
    _ax[i].set_title(name)
plt.tight_layout()
```
